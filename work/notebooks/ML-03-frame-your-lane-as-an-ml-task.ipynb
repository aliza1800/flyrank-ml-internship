## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

I will treat the Refresh / Content Opportunity Scoring problem as a ranking task. The goal is to rank content pages by how strongly the available observable signals suggest that a page should be reviewed first. Ranking fits this problem because the reviewer has limited time and needs an ordered list of pages to review, rather than only a yes/no prediction.

```python
task_type = "Ranking"

print("ML task type:", task_type)
print("Goal: rank content pages for review priority.")
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target will be a content opportunity score used to rank pages for review priority. Because there is no observed outcome showing that a refresh definitely worked, I will use a defined proxy based on available search and engagement signals. The proxy is a prioritization signal, not a ground-truth label.

```python
target_type = "Defined proxy"

print("Target:", "Content opportunity score")
print("Label source:", target_type)
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

I will use Precision@K to measure how useful the top-ranked pages are for review. A higher Precision@K means that more of the pages selected near the top meet the defined opportunity criteria. I will use a fixed K so the result matches a realistic limited review capacity.

```python
metric = "Precision@K"

print("Success metric:", metric)
print("Good means: a higher share of top-ranked pages meet the opportunity criteria.")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed if-statement would be limited because content opportunity can depend on several signals at the same time. A page may have high impressions but weak engagement, while another may have lower visibility but stronger signs of opportunity. ML can learn patterns across multiple signals and produce a ranking rather than relying on one manually chosen threshold. The result will still be treated as decision-support, not guaranteed performance.

```python
print("Why ML:", "Multiple signals interact, so a learned ranking can be more flexible than one fixed rule.")
print("Use:", "Decision-support for human content review.")
```

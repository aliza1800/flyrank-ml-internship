## 1. My rule and its reason codes

I will prioritize pages that show a combination of meaningful search visibility and signs of weak engagement or declining performance. Pages with stronger opportunity signals will receive a higher baseline score. The rule will use simple observable signals and will output reason codes so that each recommendation is easy for a human reviewer to understand.

```python
print("Rule: combine observable visibility and weakness signals.")
print("Reason codes: HIGH_IMPRESSIONS, LOW_CTR, LOW_ENGAGEMENT, DECLINING")
```

## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

```python
!git clone https://github.com/aliza1800/flyrank-ml-internship.git
```

```python
print(df.columns.tolist())
```

```python
import pandas as pd
import os

# Load the dataset
df = pd.read_csv("/content/flyrank-ml-internship/data/raw/content_refresh_anonymized.csv")

# Create a simple baseline score
df["baseline_score"] = (
    df["impressions_90d"].rank(pct=True) +
    (1 - df["ctr"].rank(pct=True)) +
    (1 - df["engagement_rate"].rank(pct=True))
)

# Rank pages
df["rank"] = df["baseline_score"].rank(ascending=False, method="first").astype(int)

# Add a simple reason code
df["reason_code"] = "HIGH_IMPRESSIONS_LOW_ENGAGEMENT"

# Create output folder
os.makedirs("/content/flyrank-ml-internship/work/outputs", exist_ok=True)

# Save ranked queue
output_path = "/content/flyrank-ml-internship/work/outputs/baseline_action_score.csv"
df.sort_values("baseline_score", ascending=False).to_csv(output_path, index=False)

print("Rows ranked:", len(df))
print("Output saved:", output_path)
print(df[["rank", "baseline_score", "reason_code"]].head(10))
```

## 3. Top-20 review

*For each of the top 20: action, reason code, confidence note, and what would make it wrong.*

The top 20 pages will be reviewed as a prioritized queue. Each page will receive an action, a reason code, a confidence note, and a note describing what could make the recommendation wrong. The recommendations are directional and intended to support human review.

```python

top20 = df.sort_values("baseline_score", ascending=False).head(20).copy()

top20["action"] = "Review for refresh"
top20["reason_code"] = "Opportunity signals"
top20["confidence_note"] = "Moderate confidence based on observed signals"
top20["what_could_make_it_wrong"] = (
    "Recent changes or missing context could change the recommendation"
)

display(top20[[
    "content_id",
    "action",
    "reason_code",
    "confidence_note",
    "what_could_make_it_wrong"
]])
```

## 4. Weak picks + leakage check

*Which picks look wrong and why? Confirm no product flags or future windows leaked in.*

The top picks are directional recommendations rather than guaranteed refresh decisions. Some picks could be wrong if recent content changes, seasonality, tracking issues, or missing context affected the observed metrics. The baseline score uses impressions_90d, ctr, and engagement_rate only. No future-window metrics or product/flag columns were used in the score, so the baseline ranking does not rely on future information.

```python
# Weak picks check
print("Top 20 baseline score range:")
print(top20["baseline_score"].min(), "to", top20["baseline_score"].max())

print("\nReason codes in top 20:")
print(top20["reason_code"].value_counts())

# Leakage check: confirm the score only uses observed 90-day metrics
score_columns = [
    "impressions_90d",
    "ctr",
    "engagement_rate"
]

print("\nColumns used in baseline score:")
print(score_columns)

# Check for product-related flags
product_columns = [
    col for col in df.columns
    if "product" in col.lower() or "flag" in col.lower()
]

print("\nProduct/flag columns found:")
print(product_columns if product_columns else "None")

print("\nLeakage check:")
print("Baseline score uses observed 90-day metrics only.")
print("No future-window columns were used in the score.")
```
